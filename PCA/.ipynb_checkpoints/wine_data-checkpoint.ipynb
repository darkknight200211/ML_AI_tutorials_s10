```python
# importing required libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# importing or loading the dataset
path_to_data    =    "/home/jwalitnpanchal/college_work/sem10/ML_AI/PCA/Wine.csv"
dataset = pd.read_csv(path_to_data)
# distributing the dataset into two components X and Y
x = dataset.iloc[:, 0:13].values    # taking the first 13 coloumns of the database
y = dataset.iloc[:, 13].values      # taking the 14th coloumn of the database
```

```python
# Splitting the X and Y into the
# Training set and Testing set
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
########## transforming the given data into standardized form #################
sc    =    StandardScaler()
x_train    =    sc.fit_transform(x_train)
x_test     =    sc.transform(x_test)

### instantiating the PCA class
pca    =    PCA(n_components=10)
x_train    =    pca.fit_transform(x_train)
x_test     =    pca.transform(x_test)

#### fit_transform ##########
x_pca    =    pca.fit_transform(x_train)

figsize = (5,3)
plt.figure(1, figsize=figsize)
plt.scatter(x_train[:,0], x_train[:,1],s=1.5, color="red")
plt.tight_layout()
plt.figure(2, figsize=figsize)
plt.scatter(x_pca[:,0], x_pca[:,8],s=1.5, color="red")
plt.tight_layout()
plt.show()
```
