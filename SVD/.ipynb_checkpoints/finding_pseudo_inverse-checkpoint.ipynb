```python
import numpy as np
```

```python
# initialize the array whose pseudo-inverse is to be calculated
x    =    np.array([[1,1],[0,1],[1,0]])
#print(x.shape)
# performing the SVD on the above matrix using the inbuilt SVD function
U, S, Vt    =    np.linalg.svd(x)
print(U,"\n", S,"\n", Vt)
```

```python
# calculating the pseudoinverse
# construct S_plus
S_plus    =    np.zeros((len(x[0,:]), len(x[:,0])))
for i in range(len(S)):
    S_plus[i,i]    =    1/S[i]
#print(S_plus)
x_plus    =    (Vt.T @ S_plus) @ U.T
print(x_plus)
```

```python
# checking if, indeed the above is pseudo-inverse of x or not
print(x_plus @ x)
```
