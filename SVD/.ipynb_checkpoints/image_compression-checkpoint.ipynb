```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# read the image as an np.ndarray using plt.imread
messi_image    =    plt.imread("/home/jwalitnpanchal/college_work/sem10/ML_AI/SVD/Lionel-Messi-2018.jpg")
#print(type(messi_image), messi_image.shape)
plt.imshow(messi_image)
```

```python
# for the 3 r,g,b channels create separate 2-d arrays. SVD will be performed on these separately and then added
r_channel    =    np.zeros((messi_image.shape[0],messi_image.shape[1]))
r_channel    =    messi_image[:,:,0]
g_channel    =    np.zeros((messi_image.shape[0],messi_image.shape[1]))
g_channel    =    messi_image[:,:,1]
b_channel    =    np.zeros((messi_image.shape[0],messi_image.shape[1]))
b_channel    =    messi_image[:,:,2]
```

```python
# SVD for the red channel:
u_r, sigma_r, vt_r    =    np.linalg.svd(r_channel)
print(sigma_r)
```

```python
# rank-1 decompostion
r_r1_decomp    =    np.zeros((messi_image.shape[0], messi_image.shape[1]))
dummy_sum_mat  =    np.zeros((messi_image.shape[0], messi_image.shape[1]))
add_terms      =    100 
for k in range(add_terms):
    for i in range(messi_image.shape[0]):
        for j in range(messi_image.shape[1]):
            dummy_sum_mat[i,j]    =    u_r[i,k] * vt_r[k,j]
    dummy_sum_mat[:,:]    =    dummy_sum_mat[:,:]/sigma_r[k]
    r_r1_decomp    =    r_r1_decomp + dummy_sum_mat
```

```python
plt.imshow(r_r1_decomp)
```
