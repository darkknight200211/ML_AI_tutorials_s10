```python
import numpy as np
```

```python
def pseudo_inverse(x):
    # performing the SVD on the above matrix using the inbuilt SVD function
    U, S, Vt    =    np.linalg.svd(x)
    # calculating the pseudoinverse
    # construct S_plus
    S_plus    =    np.zeros((len(x[0,:]), len(x[:,0])))
    for i in range(len(S)):
        S_plus[i,i]    =    1/S[i]
    #print(S_plus)
    x_plus    =    (Vt.T @ S_plus) @ U.T
    return x_plus
```

```python
# finding the solution to the system
coeff_matrix    =    np.array([[2,-1,1],
                              [1,1,-1],
                              [3,1,1],
                              [4,-1,-1],
                              [5,2,3]])
rhs_matrix    =    [1,7,10,0,5]
c_m_plus    =    pseudo_inverse(coeff_matrix)
solution    =    c_m_plus @ rhs_matrix
print(solution)
```
