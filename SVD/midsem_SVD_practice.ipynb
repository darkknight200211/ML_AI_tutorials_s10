```python
import numpy as np
```

```python
x = np.array([[3,2,2],[2,3,-2]])
print(x)
```

```python
u, sigma, vt   =   np.linalg.svd(x)
print(u,"\n",sigma,"\n",vt)
```

```python
# rank-1 decomposition of the matrix x
x_from_decom  =    np.zeros((x.shape[0],x.shape[1]))
sum_matrix    =    np.zeros((x.shape[0],x.shape[1]))
for k in range(sigma.shape[0]):
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            sum_matrix[i,j]    =    u[i,k] * vt[k,j]
    sum_matrix    =    sum_matrix/sigma[k]
    print(sum_matrix,"\n")
    x_from_decom  =    x_from_decom + sum_matrix
```
