```python
import pandas as pd
import statsmodels.api as sm
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
#### importing a dataset from kaggle
path_t0_data    =    "/home/jwalitnpanchal/college_work/sem10/ML_AI/case_studies/gold_price_forecasting_dataset.csv"
data1           =    pd.read_csv(path_t0_data)
#print(data1.info())
print(data1.head())
```

```python
###### extracting some predictors and one response
# predictors
opening_price    =    np.asarray(data1["open"].values).reshape(-1,1)
trade_volume     =    np.asarray(data1["volume"].values).reshape(-1,1)
ma_7_arr         =    np.asarray(data1["ma_7"].values).reshape(-1,1)
volatility_7_arr =    np.asarray(data1["volatility_7"].values).reshape(-1,1)
# response
daily_ret_arr    =    np.asarray(data1["daily_return"].values).reshape(-1,1)
######### stack the predictors into one matrix
predictor_stack =    np.column_stack((opening_price,trade_volume, ma_7_arr, volatility_7_arr))
print(predictor_stack)
predictor_stack       =    sm.add_constant(predictor_stack)
### split the data into parts
predictors_train, predictors_test, daily_ret_train, daily_ret_test    =    train_test_split(predictor_stack, daily_ret_arr, random_state=45, test_size=0.25)
######## train the model on the train dat:
print(daily_ret_train.shape, predictors_train.shape)
model_gold_pricing     =    sm.OLS(daily_ret_train, predictors_train)
fit_model_gold_pricing =    model_gold_pricing.fit()
print(fit_model_gold_pricing.summary())
```

```python
response_col    =    "daily_return"
predictors      =    data1.loc[:, data1.columns != response_col]
predictors      =    predictors.drop(columns="date")
predictors      =    sm.add_constant(predictors)
#print(predictors)
response        =    data1.loc[:, response_col]
### split the data into parts
predictors_train1, predictors_test1, daily_ret_train1, daily_ret_test1    =    train_test_split(predictors, response, random_state=45, test_size=0.25)

# fit the OLS model
model    =    sm.OLS(daily_ret_train1, predictors_train1)
results  =    model.fit()
print(results.summary())
```
