## Decision Tree Construction using Entropy and Information Gain

In a classification decision tree, node impurity can be measured using
\emph{entropy}, which quantifies the uncertainty associated with class labels.
For a node $S$ containing observations from $K$ classes, entropy is defined as
$$
H(S) = -\sum_{i=1}^{K} p_i \log_2 p_i,
$$
where $p_i$ denotes the proportion of observations belonging to class $i$ in
node $S$.

To select a split, the concept of \emph{information gain} is used. If a candidate
attribute $A$ partitions $S$ into subsets $S_1, S_2, \dots, S_m$, the information
gain due to the split is given by
$$
IG(S, A) = H(S) - \sum_{j=1}^{m} \frac{|S_j|}{|S|} \, H(S_j).
$$
The attribute with the highest information gain is chosen for splitting, as it
results in the greatest reduction in entropy and hence produces more homogeneous
child nodes. This greedy process is applied recursively until a stopping
criterion is satisfied.

```python
import pandas as pd
import numpy as np
```

```python
## The following data is was introduced by by J. Ross Quinlan the inventor of Decision Tree

# Create the Play Tennis dataset
data = pd.DataFrame({
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
                "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],
    "Temperature": ["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
                    "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],
    "Humidity": ["High","High","High","High","Normal","Normal","Normal",
                 "High","Normal","Normal","Normal","High","Normal","High"],
    "Wind": ["Weak","Strong","Weak","Weak","Weak","Strong","Strong",
             "Weak","Weak","Weak","Strong","Strong","Weak","Strong"],
    "Play": ["No","No","Yes","Yes","Yes","No","Yes",
             "No","Yes","Yes","Yes","Yes","Yes","No"]
})
data
```

```python
values, counts = np.unique(data["Play"], return_counts=True)
print(values)
print(counts)
```

```python
## Define Entropy Function
def entropy(labels):
    values, counts = np.unique(labels, return_counts=True)
    probabilities = counts / counts.sum()
    return -np.sum(probabilities * np.log2(probabilities))
```

```python
## Entropy of the Root Node
root_entropy = entropy(data["Play"])
root_entropy
```

```python
## Information Gain Function
def information_gain(data, feature, target="Play"):
    total_entropy = entropy(data[target])
    values = data[feature].unique()
    
    weighted_entropy = 0
    for v in values:
        subset = data[data[feature] == v]
        weighted_entropy += (len(subset)/len(data)) * entropy(subset[target])
    
    return total_entropy - weighted_entropy
```

```python
##  Compute Information Gain for Each Feature
features = ["Outlook", "Temperature", "Humidity", "Wind"]

for f in features:
    print(f, ":", information_gain(data, f))
```

```python
## Choose Best Feature (Root Node)
best_feature = max(features, key=lambda f: information_gain(data, f))
best_feature
```

```python
## Split the Dataset on the Best Feature
def split_data(data, feature):
    splits = {}
    for value in data[feature].unique():
        splits[value] = data[data[feature] == value]
    return splits
```

```python
splits = split_data(data, best_feature)
splits.keys()
```

```python
## Check for Pure Nodes (Stopping Condition)
def is_pure(subset, target="Play"):
    return len(subset[target].unique()) == 1
```

```python
for key, subset in splits.items():
    print(key, "pure:", is_pure(subset))
```

```python
## Recursive Tree Construction
def id3(data, features, target="Play"):
    
    # If all samples belong to one class --> leaf
    if len(data[target].unique()) == 1:
        return data[target].iloc[0]
    
    # If no features left --> majority class
    if len(features) == 0:
        return data[target].mode()[0]
    
    # Choose best feature
    gains = {f: information_gain(data, f, target) for f in features}
    best_feature = max(gains, key=gains.get)
    
    tree = {best_feature: {}}
    
    for value in data[best_feature].unique():
        subset = data[data[best_feature] == value]
        remaining_features = [f for f in features if f != best_feature]
        tree[best_feature][value] = id3(subset, remaining_features, target)
    
    return tree
```

```python
tree = id3(data, features)
tree
```

```python
def predict(tree, sample):
    if not isinstance(tree, dict):
        return tree
    
    feature = next(iter(tree))
    value = sample[feature]
    return predict(tree[feature][value], sample)
```

```python
sample = {
    "Outlook": "Overcast",
    "Temperature": "Mild",
    "Humidity": "High",
    "Wind": "Weak"
}

predict(tree, sample)
```

## Using Gini (index) Impurity

In a classification decision tree, splits are selected by minimizing the
**weighted Gini index** of the child nodes. For a node $S$ containing
observations from $K$ classes, the Gini impurity is defined as
$$
G(S) = 1 - \sum_{i=1}^{K} p_i^2,
$$
where $p_i$ is the proportion of class $i$ in $S$.

If a candidate split of an attribute $A$ partitions $S$ into subsets
$S_1, S_2, \dots, S_m$, the weighted Gini index after the split is
$$
G(S \mid A) = \sum_{j=1}^{m} \frac{|S_j|}{|S|} \, G(S_j).
$$
The optimal split is the one that minimizes $G(S \mid A)$, thereby producing
child nodes that are as homogeneous as possible. This greedy splitting process
is applied recursively until stopping criteria such as node purity or minimum
node size are satisfied, and each terminal node predicts the majority class of
the observations it contains.

```python
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame({
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
                "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],
    "Temperature": ["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
                    "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],
    "Humidity": ["High","High","High","High","Normal","Normal","Normal",
                 "High","Normal","Normal","Normal","High","Normal","High"],
    "Wind": ["Weak","Strong","Weak","Weak","Weak","Strong","Strong",
             "Weak","Weak","Weak","Strong","Strong","Weak","Strong"],
    "Play": ["No","No","Yes","Yes","Yes","No","Yes",
             "No","Yes","Yes","Yes","Yes","Yes","No"]
})
```

```python
def gini(labels):
    probs = labels.value_counts(normalize=True)
    return 1 - np.sum(probs**2)
```

```python
gini_root = gini(data["Play"])
gini_root
```

```python
def weighted_gini(data, feature, target="Play"):
    total = len(data)
    wg = 0
    
    for value, subset in data.groupby(feature):
        wg += (len(subset)/total) * gini(subset[target])
    
    return wg
```

```python
features = ["Outlook", "Temperature", "Humidity", "Wind"]

for f in features:
    print(f, ":", weighted_gini(data, f))
```

```python
best_root = min(features, key=lambda f: weighted_gini(data, f))
best_root
```

```python
outlook_splits = {
    value: subset
    for value, subset in data.groupby("Outlook")
}

for k, v in outlook_splits.items():
    print(k, ": Gini:", gini(v["Play"]))
```

```python
outlook_splits["Overcast"]["Play"].value_counts()
```

```python
sunny = outlook_splits["Sunny"]
gini(sunny["Play"])
```

```python
for f in ["Temperature", "Humidity", "Wind"]:
    print(f, ":", weighted_gini(sunny, f))
```

```python
for value, subset in sunny.groupby("Humidity"):
    print(value, ": Gini:", gini(subset["Play"]))
```

```python
rain = outlook_splits["Rain"]
gini(rain["Play"])
```

```python
for f in ["Temperature", "Humidity", "Wind"]:
    print(f, ":", weighted_gini(rain, f))
```

```python
for value, subset in rain.groupby("Wind"):
    print(value, ": Gini:", gini(subset["Play"]))
```

## Using sklearn

```python
import pandas as pd
import numpy as np
```

```python
# Create the Play Tennis dataset
data = pd.DataFrame({
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
                "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],
    "Temperature": ["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
                    "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],
    "Humidity": ["High","High","High","High","Normal","Normal","Normal",
                 "High","Normal","Normal","Normal","High","Normal","High"],
    "Wind": ["Weak","Strong","Weak","Weak","Weak","Strong","Strong",
             "Weak","Weak","Weak","Strong","Strong","Weak","Strong"],
    "Play": ["No","No","Yes","Yes","Yes","No","Yes",
             "No","Yes","Yes","Yes","Yes","Yes","No"]
})

data
```

```python
from sklearn.preprocessing import LabelEncoder

encoded_data = data.copy()

encoders = {}
for column in encoded_data.columns:
    le = LabelEncoder()
    encoded_data[column] = le.fit_transform(encoded_data[column])
    encoders[column] = le

encoded_data
```

```python
X = encoded_data.drop("Play", axis=1)
y = encoded_data["Play"]
```

```python
from sklearn.tree import DecisionTreeClassifier

# CART uses Gini index by default
cart_model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=None,
    random_state=0
)

cart_model.fit(X, y)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(14,8))
plot_tree(
    cart_model,
    feature_names=X.columns,
    class_names=encoders["Play"].classes_,
    filled=True
)
plt.show()
```

```python
sample = pd.DataFrame({
    "Outlook": ["Sunny"],
    "Temperature": ["Hot"],
    "Humidity": ["High"],
    "Wind": ["Weak"]
})

for col in sample.columns:
    sample[col] = encoders[col].transform(sample[col])

prediction = cart_model.predict(sample)
encoders["Play"].inverse_transform(prediction)
```

```python
from sklearn.tree import export_text

rules = export_text(cart_model, feature_names=list(X.columns))
print(rules)
```

## Example with syntetic random data 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
np.random.seed(42)

n = 200
X = np.random.randn(n, 2)

# Define a simple nonlinear decision rule
y = (X[:, 0]**2 + X[:, 1] > 1).astype(int)

df = pd.DataFrame(X, columns=["Feature_1", "Feature_2"])
df["Target"] = y

df.head()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df["Feature_1"], df["Feature_2"], c=df["Target"])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Synthetic Dataset")
plt.grid(True)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df[["Feature_1", "Feature_2"]],
    df["Target"],
    test_size=0.3,
    random_state=42
)
```

```python
dt = DecisionTreeClassifier(
    criterion="gini",
    max_depth=3,
    random_state=42
)

dt.fit(X_train, y_train)
```

```python
y_pred = dt.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
plt.figure(figsize=(14,6))
plot_tree(
    dt,
    feature_names=["Feature_1", "Feature_2"],
    class_names=["Class 0", "Class 1"],
    filled=True
)
plt.show()
```

```python
from sklearn.metrics import classification_report

y_pred = dt.predict(X_test)

print(classification_report(y_test, y_pred))
```

## Decision Tree on Cancer Dataset from sklearn

```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="Target")

X.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
model = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=4,
    min_samples_leaf=10,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
