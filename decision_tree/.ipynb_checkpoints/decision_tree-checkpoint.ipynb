```python
import numpy as np
```

```python
# entropy of full data
hs    =    -(9/14)*np.log2(9/14) - (5/14)*np.log2(5/14)
print(hs)
```

## Temp ##

```python
h_hot    =    (-2/4)*np.log2(2/4) - (2/4)*np.log2(2/4)
h_mild   =    (-4/6)*np.log2(4/6) - (2/6)*np.log2(2/6)
h_cold   =    (-3/4)*np.log2(3/4) - (1/4)*np.log2(1/4)
print(f"cond. entropy hot {h_hot:.3e}\n"
     f"cond. entropy mild {h_mild:.3e}\n"
f"cond. entropy cold {h_cold:.3e}\n")
```

```python
cond_entropy_temp    =    (4/14)*h_hot + (6/14)*h_mild + (4/14)*h_cold
print(cond_entropy_temp)
```

```python
info_gain_temp    =    hs - cond_entropy_temp
print(info_gain_temp)
```
