```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
# generating synthetic data
from sklearn.datasets import make_blobs
number_of_cluster    =    4
num_features    =    2
x_data, y_data, centers    =    make_blobs(n_samples=100,
    n_features=num_features,
    centers=number_of_cluster,
    cluster_std=1.0,
    center_box=(-10.0, 10.0),
    shuffle=True,
    random_state=53,
    return_centers=True,
)
print(x_data, "\n", y_data, "\n", centers)
```

```python
# plotting the initial data
list_labels   =    np.unique(y_data)
print(list_labels)
color_list   =   ["red", "blue", "green","brown"]
figsize=(5,3)
plt.figure(1,figsize=figsize)
for i in list_labels:
    uni_data    =    x_data[y_data==i]
    plt.scatter(uni_data[:,0],uni_data[:,1],s=3.5,c=color_list[i])
    plt.scatter(centers[i,0],centers[i,1],s=42,c="purple",marker="*")
plt.show()
```

```python
# standardize the input data
std_x_data    =    StandardScaler()
std_transformed_data    =    std_x_data.fit_transform(x_data)
print(std_transformed_data,"\n",
      f"mean of c1 =\t{np.mean(std_transformed_data[:,0]):.3f}\n",
      f"mean of c2 =\t{np.mean(std_transformed_data[:,1]):.3f}\n", 
        f"std of c1 =\t{np.std(std_transformed_data[:,0]):.3f}\n",
        f"std of c2 =\t{np.std(std_transformed_data[:,1]):.3f}")
```

```python
# k-means clustering
k_means    =    KMeans(n_clusters=number_of_cluster,random_state=41,max_iter=300)
clustered_transformed_data    =    k_means.fit(std_transformed_data)
labels    =    clustered_transformed_data.labels_
centers_kmeans    =    clustered_transformed_data.cluster_centers_
print(labels,"\n",centers_kmeans,"\n",type(clustered_transformed_data))
```

```python
availabel_labels    =    np.unique(labels)
print(availabel_labels)
for i in availabel_labels:
    #unique_clustered_data    =    clustered_transformed_data[i]
    plt.scatter(std_transformed_data[:,0],std_transformed_data[:,1],s=10,c=labels)
    plt.scatter(centers_kmeans[i,0],centers_kmeans[i,1],s=50,marker="*",color="purple")
```
