```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
data_directory_path    =    "/home/jwalitnpanchal/college_work/sem10/ML_AI/k_means_clustering/iris/iris.data"
iris_data_raw    =    pd.read_csv(data_directory_path).drop(columns="Iris-setosa")
print(iris_data_raw)
```

```python
# standardize the data
std_scalar    =    StandardScaler()
iris_data_std =    std_scalar.fit_transform(iris_data_raw)
print(iris_data_std)
```

```python
# kmeans clustering
kmeans_class    =    KMeans(n_clusters=4,random_state=51,max_iter=300)
clustered_iris_data    =    kmeans_class.fit(iris_data_std)
labels    =    clustered_iris_data.labels_
cluster_centers    =    clustered_iris_data.cluster_centers_
clustered_iris_data    =    np.asarray(clustered_iris_data)
print(labels, "\n", cluster_centers)
```

```python
# plotting the data
availabel_labels    =    np.unique(labels)
colorlist    =    ["red", "green", "blue", "black"]
for i in availabel_labels:
    #unique_clut_data    =    clustered_iris_data[labels==i]
    plt.scatter(iris_data_std[:,1],iris_data_std[:,2],s=5,c=labels)
    plt.scatter(cluster_centers[i,1],cluster_centers[i,2],s=50,marker="*",color="purple")
plt.show()
```
